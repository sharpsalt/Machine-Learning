```python
class PercentileDetector:
  def __init__(self,percentile=0.95):  #this is default value isko humlog apne hisaab se change kr kste hai
    self.percentile=percentile
  def fit(self,X,y=None):
    self.thresholds=[
        pd.Series(X[:,i]).quantile(self.percentile) for i in range(X.shape[1])
    ]
  def predict(self,X,y=None):
    return (X>self.thresholds).max(axis=1)
  def fit_predict(self,X,y=None):
    self.fit(X)
    return self.predict(X)
```

## Elliptical Envelope

```python
#We are going to put Ellipse on the dataset, and using that ellipse like anything which is falling beyondellipse is going to be my Anomalies
```

```python
from sklearn.covariance import EllipticEnvelope
```

```python
ee=EllipticEnvelope(random_state=0) # we would be able to repeatdely find something
y_pred=ee.fit_predict(X)
```

```python
#all these methods are very similar to dbscan , when we run dbscan on data it forms clsuters and anythin it couldn't form a cluster (outlier) to usko -1 krdeta hai yaha pe bhi same hi hai
```

```python
y_pred
```

```python
# y_pred from EllipticEnvelope: 1 for inliers (normal), -1 for outliers (anomalies)
# y (true labels): 0 for normal, 1 for anomalies

# Map y_pred to match y: 1 -> 0 (normal), -1 -> 1 (anomaly)
y_pred_mapped = np.where(y_pred == -1, 1, 0)
metrics_(y, y_pred_mapped)
```

```python
#here our precision is only 9%, while the recall is 45%
plt.figure(figsize=(10,8))
sns.scatterplot(x=X[:,0],y=X[:,1],hue=y_pred,style=y)
```

```python
#In out original data hum dekhrhe the ki usme bahut kam anomalies tha isme kafi different hai
#like some fo them have picked up by our elliptical envelope and too many false positive
#So thta's why recall was better but precision is still verybad
```

## Local Outlier Factor(LOF)

```python
from sklearn.neighbors import LocalOutlierFactor
# loc=LocalOutlierFactor()
#Most of the paramaters are known to us because we ahve learn clustering and other techniques
```

```python
#This model is going to use some sort of neighbors to detect anomaly or not
#Maybe the simplest explanation
#If you have a lot of observation in neighbourhood then it is definately not anomaly
#If you have a very less observation in neighbourhoodthen it is marked as anomaly(low density area)
```

```python
#One of the main factor is neighbors, and typically it uses Distnace for the calculation of it
#if we have p=1 that mean we are using Manhattan distance and p=2 means Euclidean Distance
```

```python
Loc=LocalOutlierFactor(n_neighbors=20)
y_pred=Loc.fit_predict(X)==-1
```

```python
metrics_(y,y_pred)
#n_neighbors=50 pe mera precision=36% and recall was 65% tha
#but jab hum n_neighbors=20 kiye to mera precsion 34% and recall 65% hogaya
```

```python
plt.figure(figsize=(10,8))
sns.scatterplot(x=X[:,0],y=X[:,1],hue=y_pred,style=y)
```

```python
#it uses monkowski as metrics
```

```python
Loc_n=LocalOutlierFactor(n_neighbors=20,metric='manhattan',novelty=True)
#jab hum novelty=True karenge then that means all of these obsrvation which are anomalies won;t be used for trainigw while the normal one were used to decide the density in the area
#Now when we pass new dataset it should predict
```

```python
#You are actually trying to learn Noevelty=True
#Then we give
Loc_n.fit(X)
```

```python
y_pred=Loc_n.predict(X)
```

```python
y_pred_mapped = np.where(y_pred == -1, 1, 0) # Remap y_pred: -1 (outlier) to 1 (anomaly), 1 (inlier) to 0 (normal)
metrics_(y, y_pred_mapped)
```

```python
#Recall styaed same and precisoon imporved by avery little bit
plt.figure(figsize=(10,10))
sns.scatterplot(x=X[:,0],y=X[:,1],hue=y_pred,style=y)
```
